## Problem 1

#### Distinct

```python
# Creating the function
def solution(A):

    # Creating a list to track the unique values we encounter
    unique_values = []
    
    # Looping over A and checking if each value has been seen before
    for idx in range(len(A)):
        if A[idx] not in unique_values:
            unique_values.append(A[idx])
    
    # Returning the final answer
    return len(unique_values)

# Testing the function with the example from the question
A = [2, 1, 1, 2, 3, 1]
solution(A)
```

## Problem 2

#### MaxProductOfThree

```python
# Creating the function
def solution(A):

    # Checking that there are at least 3 elements in list A
    if len(A) < 3:
        print("Please input a list containing at least 3 elements.")
        return None

    # The maximum product comes from one of two cases:
    #   1) the 3 largest numbers (e.g. 3 positives), or
    #   2) the 2 smallest (most negative) numbers times the largest number
    #      (2 negatives make a positive, times 1 positive).
    # Any other combination of signs gives a non-positive product. That only
    # matters when no positive product is possible (e.g. all numbers negative),
    # and then the best product is still the product of the 3 largest numbers.

    # Using selection sort to place the 2 smallest numbers on the left and 3 largest numbers 
    # on the right (the 3rd through n-3rd numbers don't need to be sorted by the logic above)

    # Placing the two smallest elements on the left
    i = 0
    while i < 2:
        min_val = i
        j = i + 1
        while j < len(A):
            if A[j] < A[min_val]:
                min_val = j
            j += 1
        if min_val != i:
            A[i], A[min_val] = A[min_val], A[i]
        i += 1
        
    # Placing the three largest elements on the right
    # (don't need to check index 0 and 1 as they are the smallest elements)
    i = len(A)-1
    while i > (len(A)-1)-3:
        max_val = i
        j = i - 1
        while j > 1:
            if A[j] > A[max_val]:
                max_val = j
            j -= 1
        if max_val != i:
            A[i], A[max_val] = A[max_val], A[i]
        i -= 1
        
    # Computing the possible solutions (3 positive or 1 positive and 2 negative)
    candidate_solution_1 = A[-3] * A[-2] * A[-1]
    candidate_solution_2 = A[0] * A[1] * A[-1]

    # Choosing the maximum solution
    max_product = max(candidate_solution_1, candidate_solution_2)

    # Returning the final answer
    return max_product

# Testing the function with the example from the question
A = [-3, 1, 2, -2, 5, 6]
solution(A)
```

## Problem 4

#### AbsDistinct

```python
# Creating the function
def solution(A):

    # Re-sorting the list from largest to smallest absolute value by
    # taking the largest of the absolute values from the values at 
    # either end of the list and adding it to our new sorted list
    # Since A is sorted, one of the ends must have the largest absolute value
    
    # Creating tracking variables
    sorted_abs = []
    i = 0
    j = len(A) - 1

    # Sorting the new list
    for elements in range(len(A)):
        if abs(A[i]) < abs(A[j]):
            sorted_abs.append(abs(A[j]))
            j -= 1
        else:
            sorted_abs.append(abs(A[i]))
            i += 1

    # Checking for distinct values
    # (starting the count at 1 since the first element is distinct by definition)
    abs_distinct = 1
    for element in range(1, len(A)):
        if sorted_abs[element] != sorted_abs[element-1]:
            abs_distinct += 1

    # Returning the final answer
    return abs_distinct

# Testing the function with the example from the question
A = [-5, -3, -1, 0, 3, 6]
solution(A)
```

## Problem 6

#### CountTriangles

```python
# Creating the function
def solution(A):

    # Using selection sort to order the numbers from largest to smallest
    i = 0
    while i < len(A):
        max_val = i
        j = i + 1
        while j < len(A):
            if A[j] > A[max_val]:
                max_val = j
            j += 1
        if max_val != i:
            A[i], A[max_val] = A[max_val], A[i]
        i += 1

    # After sorting from largest to smallest, take any triplet with indices i < j < k.
    # Then A[i] >= A[j] >= A[k], so A[i] is the longest side. All sides are positive,
    # so A[i] + A[j] > A[k] and A[i] + A[k] > A[j] are always true, because A[i] is
    # already at least as big as A[j] and A[k]. The only condition left to check is
    # A[j] + A[k] > A[i]. So I fix i as the longest side and test every pair j < k
    # that comes after it.

    # Looping over each index i, j, and k
    triangles = 0
    for i in range(len(A)-2):
        for j in range(i+1, len(A)-1):
            for k in range(j+1, len(A)):
                if A[i] < A[j] + A[k]:
                    triangles += 1
                
                # If A[i] >= A[j] + A[k] for some k, then since the list is sorted
                # already, the rest of the elements will also not form a triangle
                else:
                    break
            
            # If A[i] >= A[j] + A[j+1], then even the best partner for j (the next 
            # element) can't beat A[i]. Every later j has smaller values, so no later 
            # j can form a triangle with i either, and we can stop the j loop.
            if A[i] >= A[j] + A[j+1]:
                break
        
    # Returning the final answer
    return triangles

# Testing the function with the example from the question
A = [10, 2, 5, 1, 8, 12]
solution(A)
```

## Problem 7

#### NumberOfDiscIntersections

```python
# Creating the function
def solution(A):

    # Calculating where the left and right edges of each circle are
    left_edges = []
    right_edges = []
    for idx in range(len(A)):
        left_edges.append(idx - A[idx])
        right_edges.append(idx + A[idx])

    # Any pair of disks does not intersect if they are disjoint, therefore, if
    # the value of its right edge is less than the left edge of another disk or the
    # value of its left edge is greater than the right edge, they don't intersect

    # Counting the number of intersections and adding stopping condition
    intersections = 0
    for disk in range(len(A)):
        
        # (only counting disks with a greater index than the one 
        #  we are checking to prevent double counting)
        for comp in range(disk + 1, len(A)):
            if intersections > 10000000:
                break
            if right_edges[disk] < left_edges[comp] or left_edges[disk] > right_edges[comp]:
                continue
            else:
                intersections += 1

    # Checking if the stopping condition was met
    if intersections > 10000000:
        intersections = -1
    
    # Returning the final answer
    return intersections

# Testing the function with the example from the question
A = [1, 5, 2, 1, 4, 0]
solution(A)
```
