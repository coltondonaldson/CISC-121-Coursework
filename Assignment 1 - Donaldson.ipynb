## Problem 3

#### FrogJmp

```python
# Creating the function
def solution(X, Y, D):

    # Finding the necessary values
    distance_from_goal = Y-X
    jump_distance = D

    # Calculating the number of full jumps that can be made from X without going past Y
    jumps_needed = distance_from_goal // jump_distance

    # Adding the final jump needed to get past the threshold of Y
    # (Don't need to add one if D evenly divides Y-X since the final full jump 
    #  will land directly on Y)
    if distance_from_goal % jump_distance != 0:
        jumps_needed += 1

    # Returning the final answer
    return jumps_needed

# Testing the function with the example from the question
X = 10
Y = 85
D = 30
solution(X, Y, D)
```

## Problem 4

#### PermMissingElem

```python
# Creating the function
def solution(A):

    # Defining tracking variables
    elements_found = []
    for i in range(len(A)+1):
        elements_found.append("Not Found")

    # Looping over each element of A and recording each number we see
    for num in range(len(A)):
        elements_found[A[num]-1] = "Found"

    # Finding the index of the number we have not found yet
    # (and adding 1 since indicies begin at 0 and our array is from 1 to N+1)
    missing_element_index = elements_found.index("Not Found")
    missing_element = missing_element_index + 1

    # Returning the final answer
    return missing_element

# Testing the function with the example from the question
A = [2, 3, 1, 5]
solution(A)
```

## Problem 7

#### FrogRiverOne

```python
# Creating the function
def solution(X, A): 

    # Defining counting variables
    # (time passed starts at -1 since the first leaf falls at time 0, 
    #  and I add 1 to the time for each leaf that falls)
    time_passed = -1
    leaves_in_each_position = []
    for i in range(X):
        leaves_in_each_position.append(0)

    # Looping over each element of A until there is at least 1 leaf in each position
    for index in range(len(A)):
        if min(leaves_in_each_position) == 0:
            
            # Tracking the time passed and the adding the leaf to the positional counter
            time_passed += 1
            leaves_in_each_position[A[index]-1] += 1
        
        # Ending the loop if there is at least 1 leaf in each position
        else:
            break

    # Checking to see if the frog is able to cross the river after all leaves fall
    if min(leaves_in_each_position) == 0:
        time_passed = -1

    # Returing the final answer
    return time_passed

# Testing the function with the example from the question
X = 5
A = [1, 3, 1, 4, 2, 3, 5, 4]
solution(X, A)
```

## Problem 8

#### PassingCars

```python
# Creating the function
def solution(A):

    # Defining counting variables
    pairs_of_passing_cars = 0
    number_of_1s_seen = 0

    # Looping over each element of A, from back to front
    for index in range(len(A)):
        if A[len(A)-index-1] == 1:
            number_of_1s_seen += 1
        elif A[len(A)-index-1] == 0:
            pairs_of_passing_cars += number_of_1s_seen

        # Checking if the upper limit has been reached
        if pairs_of_passing_cars > 1000000000:
            pairs_of_passing_cars = -1
            break
    
    # Returning the final answer
    return pairs_of_passing_cars

# Testing the function with the example from the question
A = [0, 1, 0, 1, 1]
solution(A)
```

## Problem 10

#### MaxCounters

```python
# Creating the function
def solution(N, A):

    # Initializing the counters with a length N array of zeros
    counters = []
    for i in range(N):
        counters.append(0)

    # Looping over each element of A
    for index in range(len(A)):

        # Setting all counters equal to the max if A[K] = N+1
        if A[index] == N+1:
            current_max = max(counters)
            for j in range(N):
                counters[j] = current_max
        
        # Increasing counter A[K] by one if A[K] = X such that 1<=X<=N
        elif A[index] < N+1:
            counters[A[index]-1] += 1

    # Returning the final answer
    return counters

# Testing the function with the example from the question
N = 5
A = [3, 4, 4, 6, 1, 4, 4]
solution(N, A)
```
